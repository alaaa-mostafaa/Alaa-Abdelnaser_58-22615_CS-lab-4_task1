```python
import pandas as pd
```

```python

df = pd.read_csv("club_signups.csv")


print("First 5 rows:")
display(df.head())
```

# Look First

```python

print("Shape (rows, columns):", df.shape)

print("\nDataFrame Info:")
print(df.info())


print("\nMissing values:")
print(df.isna().sum())

```

```python
categorical_cols = ["faculty", "club", "city", "fee_paid"]

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

```python
exact_duplicates = df.duplicated().sum()

print("Exact duplicate rows:", exact_duplicates)

print("\nExact duplicate rows:")
display(df[df.duplicated(keep=False)])

df = df.drop_duplicates().copy()

print("Rows after removing exact duplicates:", len(df))
```

```python
# Compare duplicate detection before standardizing club names

duplicates_before_club_cleaning = df.duplicated(
    subset=["student_id", "club"]
).sum()

print(
    "Repeated student + club sign-ups before club cleaning:",
    duplicates_before_club_cleaning
)
```

# Faculty, club, city. 

```python

print(df["faculty"].value_counts(dropna=False))

```

```python
df["faculty"] = df["faculty"].str.strip().str.lower()


faculty_map = {
    "met": "MET",
    "media engineering and technology": "MET",
    "iet": "IET",
    "information engineering and technology": "IET",
    "ems": "EMS",
    "engineering and materials science": "EMS",
    "pharmacy": "Pharmacy",
    "pharma": "Pharmacy",
    "management": "Management"
}

df["faculty"] = df["faculty"].map(faculty_map)


print(df["faculty"].value_counts(dropna=False))

allowed_faculties = {"MET", "IET", "EMS", "Pharmacy", "Management"}

assert df["faculty"].notna().all()
assert set(df["faculty"].unique()) == allowed_faculties
```

```python
print(df["club"].value_counts(dropna=False))
```

```python
club_map = {
    "robotics": "Robotics",
    "music": "Music",
    "chess": "Chess",
    "chess club": "Chess",
    "football": "Football",
    "soccer": "Football",
    "debate": "Debate",
    "debate club": "Debate",
    "debating": "Debate"
}


df["club"] = df["club"].str.strip().str.lower().map(club_map)

print(df["club"].value_counts(dropna=False))

allowed_clubs = {"Robotics", "Debate", "Music", "Football", "Chess"}

assert df["club"].notna().all()
assert set(df["club"].unique()) == allowed_clubs
```

```python

print(df["city"].value_counts(dropna=False))
```

```python
df["city"] = df["city"].str.strip().str.lower()


city_map = {
    "cairo": "Cairo",
    "alexandria": "Alexandria",
    "alex": "Alexandria",
    "giza": "Giza",
    "el giza": "Giza"
}

df["city"] = df["city"].map(city_map)

print(df["city"].value_counts(dropna=False))

allowed_cities = {"Cairo", "Giza", "Alexandria"}

assert df["city"].notna().all()
assert set(df["city"].unique()) == allowed_cities
```

# Names and emails

```python
print(df["full_name"].value_counts(dropna=False))
```

```python
df["full_name"] = (
    df["full_name"]
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
    .str.title()
)
```

```python
print(df["full_name"].value_counts(dropna=False))
```

```python
print(df["email"].value_counts(dropna=False))
```

```python
df["email"] = df["email"].str.strip().str.lower()
```

```python
print(df["email"].value_counts(dropna=False))
```

# fee paid to boolean

```python

print(df["fee_paid"].value_counts(dropna=False))
```

```python

df["fee_paid"] = df["fee_paid"].astype(str).str.strip().str.lower()

fee_map = {
    "yes": True,
    "y": True,
    "true": True,
    "1": True,
    "no": False,
    "n": False,
    "false": False,
    "0": False
}

df["fee_paid"] = df["fee_paid"].map(fee_map)


print(df["fee_paid"].value_counts(dropna=False))
print("\nData type:", df["fee_paid"].dtype)

assert df["fee_paid"].notna().all()
assert df["fee_paid"].map(lambda x: isinstance(x, bool)).all()
```

# Date Time

```python
print(df["signed_up_at"].value_counts())
```

```python
print(df["signed_up_at"].unique())
```

```python

format1 = df["signed_up_at"].str.match(r"\d{4}-\d{2}-\d{2}")
format2 = df["signed_up_at"].str.match(r"\d{2}/\d{2}/\d{4}")


df.loc[format1, "signed_up_at"] = pd.to_datetime(
    df.loc[format1, "signed_up_at"],
    format="%Y-%m-%d %H:%M"
)

df.loc[format2, "signed_up_at"] = pd.to_datetime(
    df.loc[format2, "signed_up_at"],
    format="%d/%m/%Y %H:%M"
)

df["signed_up_at"] = pd.to_datetime(df["signed_up_at"])

print(df["signed_up_at"].head())
print("\nData type:", df["signed_up_at"].dtype)
```

```python
print(df["signed_up_at"].unique())
```

# Duplicates

```python
exact_duplicates = df.duplicated().sum()
print("Exact duplicate rows:", exact_duplicates)

print("\nExact duplicate rows:")
display(df[df.duplicated(keep=False)])
```

Exact duplicates were removed before any normilsation and turns out they were the only ones

```python
duplicates_after_club_cleaning = df.duplicated(
    subset=["student_id", "club"]
).sum()

print(
    "Repeated student + club sign-ups after club cleaning:",
    duplicates_after_club_cleaning
)

print(
    "Additional duplicates found because of club standardization:",
    duplicates_after_club_cleaning - duplicates_before_club_cleaning
)

duplicate_pairs = df[
    df.duplicated(
        subset=["student_id", "club"],
        keep=False
    )
].sort_values(
    ["student_id", "club", "signed_up_at"]
)

display(duplicate_pairs)
```

```python
before_repeat_removal = len(df)

df = (
    df.sort_values("signed_up_at")
      .drop_duplicates(
          subset=["student_id", "club"],
          keep="last"
      )
      .copy()
)

repeated_removed = before_repeat_removal - len(df)

print("Repeated sign-ups removed:", repeated_removed)
print("Rows remaining:", len(df))

assert not df.duplicated(
    subset=["student_id", "club"]
).any()

print("Check passed: student_id + club is unique.")
```

```python
assert not df.duplicated(
    subset=["student_id", "club"]
).any()

print("Check passed: student_id + club is unique.")
```

```python
assert set(df["faculty"].unique()) == {
    "MET", "IET", "EMS", "Pharmacy", "Management"
}

assert set(df["club"].unique()) == {
    "Robotics", "Debate", "Music", "Football", "Chess"
}

assert set(df["city"].unique()) == {
    "Cairo", "Giza", "Alexandria"
}

assert df["fee_paid"].dtype == bool

assert pd.api.types.is_datetime64_any_dtype(df["signed_up_at"])

print("All cleaning checks passed.")
```
